## Research Question

Is there a statistically significant correlation between ICE arrests and reported rates of rape, murder, robbery, and assault in the same month or the following month in Texas from 2019 to 2024?

## Background and Prior Work

Immigration and Customs Enforcement (ICE) is a federal agency acting under the Department of Homeland Security. The main responsibilities of ICE include enforcing immigration law, preventing terrorism, and stopping illegal movement of people and goods within the United States. 

The motivation for exploring our topic lies in the dramatic increase in funding that ICE has received with the One Big Beautiful Bill Act. Prior to the passage of the One Big Beautiful Bill Act, ICE received about 10 dollars billion in budget per year. After passing, funding increased to 85 billion dollars. With the marked uptick in funding, activity, and media coverage, ICE has become a hotly contested topic for the American public in recent months, especially with notable incidents involving Renee Good and Alex Pretti. We will attempt to gain a better understanding of the efficacy of ICE on reducing violent and property crime.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

An article published in the The Journal of Law & Economics by Thomas J. Miles and Adam B. Cox examined the relationship between immigration enforcement and local US crime rates. The research specifically investigates the effect of the Secure Communities program. The Secure Communities program uses fingerprint-sharing between local law enforcement and federal immigration authorities to facilitate detention and removal of noncitizens. This is not a direct correlate to ICE, however, it may serve as a proxy to understand whether or not strict enforcement of immigration policy reduces crime. The conclusion of the study showed that the Secure Communities program led to hundreds of thousands of detentions, but did not reduce crime-rates as expected.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

The article Immigration and Customs Enforcement Raids the Pillar of a Community: Student Achievement, Absenteeism, and Mobility Following a Large Worksite Enforcement Operation in North Texas looks at the effects that ICE had through various time extents. It serves as a point of interest as to why Texas was chosen for this project, as well as a way to question what sort of empirical and statistical impact ICE had.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

Of all states affected by ICE, Texas stands out in the number and scale of people detained. According to data from the Transactional Records Access Clearinghouse (TRAC) Immigration, there are 18,684 people in detention, as of January 22, 2026, which makes Texas the state with the most ICE detainees followed by Louisiana, 8,207 people, and California, 6,422. Geographically, Texas shares the longest border between the U.S. and Mexico, stretching 382 miles, with more than 40 border security stations across 5 sectors. The sheer size of the border and affiliated border control stations are key in understanding the subsequent size of detainees in Texas, which should be considered when examining the correlation between violent crime and ICE arrests. While Texas holds the greatest number of detainees, it is critical to understand that detention location is not the same as arrest location. Texas maintains some of the largest ICE detention centers in the country, where those arrested by ICE both in and out of Texas may be placed into. ## Hypothesis


There will be no statistically significant relationship between the level of ICE arrests in a given month and the rates of rape, murder, robbery, and aggravated assault reported in Texas during that **same month or the following month** (0-month and 1-month lag). We focus on same-month and one-month-lagged comparisons because administrative arrests are recorded at the time of apprehension; if enforcement activity deters crime, an effect would most plausibly appear within a short window. This hypothesis is consistent with prior research on the Secure Communities program, which found no significant crime-reduction effect despite large-scale detentions (Miles & Cox, 2014).

```python
# Build annual summary table
ice_df['year'] = ice_df['datetime'].dt.year
ice_yearly = ice_df.groupby('year').size().rename('ice_arrests')

annual_rows = []
for crime_type in CRIME_TYPES:
    subset = crime_df[crime_df['offense_code'] == crime_type]
    crime_yearly = subset.groupby('data_year').size().rename('crime_count')
    combined = pd.DataFrame({'ice_arrests': ice_yearly, 'crime_count': crime_yearly}).dropna()
    r_full, p_full = stats.pearsonr(combined['ice_arrests'], combined['crime_count'])
    # exclude COVID years
    combined_nc = combined[~combined.index.isin([2020, 2021])]
    if len(combined_nc) >= 3:
        r_nc, p_nc = stats.pearsonr(combined_nc['ice_arrests'], combined_nc['crime_count'])
    else:
        r_nc, p_nc = float('nan'), float('nan')
    annual_rows.append({
        'Crime Type': crime_type,
        'r (full)': round(r_full, 3),
        'p (full)': round(p_full, 3),
        'r (excl. COVID)': round(r_nc, 3) if not np.isnan(r_nc) else '—',
        'p (excl. COVID)': round(p_nc, 3) if not np.isnan(p_nc) else '—',
    })

corr_table = pd.DataFrame(annual_rows)
print("Table 1 — Annual Pearson Correlations: ICE Arrests vs. Violent Crime")
print(corr_table.to_string(index=False))

# Scatter plot grid
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
axes = axes.flatten()

for ax, crime_type, color in zip(axes, CRIME_TYPES, CRIME_COLORS):
    subset = crime_df[crime_df['offense_code'] == crime_type]
    crime_yearly = subset.groupby('data_year').size().rename('crime_count')
    combined = pd.DataFrame({'ice_arrests': ice_yearly, 'crime_count': crime_yearly}).dropna()

    covid_mask = combined.index.isin([2020, 2021])
    ax.scatter(combined.loc[~covid_mask, 'ice_arrests'], combined.loc[~covid_mask, 'crime_count'],
               color=color, s=80, zorder=3, label='Non-COVID years')
    ax.scatter(combined.loc[covid_mask, 'ice_arrests'], combined.loc[covid_mask, 'crime_count'],
               color='gold', edgecolors='gray', s=80, zorder=3, label='COVID years (2020–21)')

    # regression line (full period)
    m, b, *_ = stats.linregress(combined['ice_arrests'], combined['crime_count'])
    x_line = np.linspace(combined['ice_arrests'].min(), combined['ice_arrests'].max(), 100)
    ax.plot(x_line, m * x_line + b, color='gray', linestyle='--', linewidth=1.2, label='OLS (full)')

    for year, row in combined.iterrows():
        ax.annotate(str(year), (row['ice_arrests'], row['crime_count']),
                    fontsize=8, ha='right', va='bottom')

    row_data = corr_table[corr_table['Crime Type'] == crime_type].iloc[0]
    ax.set_title(f"{crime_type}\nr={row_data['r (full)']} (full), r={row_data['r (excl. COVID)']} (no COVID)", fontsize=9)
    ax.set_xlabel('Annual ICE Arrests')
    ax.set_ylabel('Annual Offense Count')
    ax.legend(fontsize=7)

plt.suptitle('Fig 2 — Annual ICE Arrests vs. Violent Crime (2019–2024)', fontsize=13, y=1.01)
plt.tight_layout()
plt.savefig('figures/fig2_annual_scatter.png', bbox_inches='tight')
plt.show()
```

```python
# ── 3B: Lag structure — CCF for each crime type ──────────────────────────
from statsmodels.tsa.stattools import ccf as sm_ccf

# Build monthly aligned series
monthly_ice_series = ice_df.set_index('datetime').resample('ME').size()
monthly_ice_series.index = monthly_ice_series.index.to_period('M').to_timestamp()

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
axes = axes.flatten()

MAX_LAG = 8
sig_level = 1.96 / np.sqrt(len(monthly_ice_series))  # ~95% CI band

for ax, crime_type, color in zip(axes, CRIME_TYPES, CRIME_COLORS):
    subset = monthly_df[monthly_df['Offense'] == crime_type]
    crime_series = subset.groupby('Month')['Count'].sum()
    crime_series.index = pd.to_datetime(crime_series.index).to_period('M').to_timestamp()

    merged = pd.DataFrame({
        'ice': monthly_ice_series,
        'crime': crime_series
    }).dropna()

    # Cross-correlation: positive lag k means crime at t+k vs ice at t
    ccf_vals = sm_ccf(merged['ice'], merged['crime'], nlags=MAX_LAG, adjusted=False)

    lags = np.arange(len(ccf_vals))
    ax.bar(lags, ccf_vals, color=[color if abs(v) > sig_level else 'lightgray' for v in ccf_vals])
    ax.axhline(sig_level,  color='black', linestyle='--', linewidth=0.9, label='95% CI')
    ax.axhline(-sig_level, color='black', linestyle='--', linewidth=0.9)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.set_xticks(lags)
    ax.set_xlabel('Lag (months)')
    ax.set_ylabel('Cross-correlation')
    ax.set_title(crime_type, fontsize=10, fontweight='bold')
    ax.legend(fontsize=7)
    ax.set_ylim(-0.6, 0.6)

plt.suptitle('Fig 4 — Cross-Correlation: ICE Arrests (t) vs. Crime Count (t + lag)', fontsize=12, y=1.01)
plt.tight_layout()
plt.savefig('figures/fig4_ccf.png', bbox_inches='tight')
plt.show()
print(f"95% significance band: ±{sig_level:.3f}")
```

```python
# ── 3C: Formal Pearson tests — lag 0 and lag 1 ───────────────────────────
monthly_ice_df = monthly_ice_series.reset_index()
monthly_ice_df.columns = ['Month', 'ice_arrests']

results = []

for lag in [0, 1, 2]:
    for crime_type in CRIME_TYPES:
        subset = monthly_df[monthly_df['Offense'] == crime_type]
        crime_monthly = subset.groupby('Month')['Count'].sum().reset_index()
        crime_monthly.columns = ['Month', 'crime_count']

        # Shift crime forward by lag (crime at t+lag vs ice at t)
        crime_monthly_shifted = crime_monthly.copy()
        crime_monthly_shifted['Month'] = crime_monthly_shifted['Month'] - pd.DateOffset(months=lag)

        merged = pd.merge(monthly_ice_df, crime_monthly_shifted, on='Month')

        r, p = stats.pearsonr(merged['ice_arrests'], merged['crime_count'])
        results.append({
            'Crime Type': crime_type,
            'Lag (months)': lag,
            'n': len(merged),
            'r': round(r, 3),
            'p-value': round(p, 4),
            'Significant (p<0.05)': 'Yes' if p < 0.05 else 'No'
        })

results_df = pd.DataFrame(results)
print("Table 2 — Monthly Pearson Correlations: ICE Arrests vs. Violent Crime")
print(results_df.to_string(index=False))
```

```python
# ── 3D: Sensitivity — exclude COVID months (2020–2021) ───────────────────
covid_months = pd.date_range('2020-01-01', '2021-12-01', freq='MS')

sensitivity_results = []

for lag in [0, 1]:
    for crime_type in CRIME_TYPES:
        subset = monthly_df[monthly_df['Offense'] == crime_type]
        crime_monthly = subset.groupby('Month')['Count'].sum().reset_index()
        crime_monthly.columns = ['Month', 'crime_count']

        crime_monthly_shifted = crime_monthly.copy()
        crime_monthly_shifted['Month'] = crime_monthly_shifted['Month'] - pd.DateOffset(months=lag)

        merged = pd.merge(monthly_ice_df, crime_monthly_shifted, on='Month')
        # Exclude COVID period
        merged_nc = merged[~merged['Month'].isin(covid_months)]

        r_full, p_full = stats.pearsonr(merged['ice_arrests'], merged['crime_count'])
        r_nc, p_nc = stats.pearsonr(merged_nc['ice_arrests'], merged_nc['crime_count'])

        sensitivity_results.append({
            'Crime Type': crime_type,
            'Lag': lag,
            'r (full)': round(r_full, 3),
            'p (full)': round(p_full, 4),
            'r (no COVID)': round(r_nc, 3),
            'p (no COVID)': round(p_nc, 4),
        })

sens_df = pd.DataFrame(sensitivity_results)
print("Table 3 — Sensitivity Analysis: Full Period vs. Excluding COVID (2020–2021)")
print(sens_df.to_string(index=False))
```
